This notebook contains a few extra functions modified to make use of MLX arrays instead of numpy ones. I'm not sure if any of these are actually needed. I'm mostly using this notebook to get a little more familiar with MLX in general.

```python
import mlx.core as mx
```

```python
# The test array

a = mx.array([1, 4, 3])
print(a)
```

```python
def max_mx(array):
    """
    Returns the maximum value of an array.
    """
    if array.dtype == 'mlx.core.complex64':
        return 'Error: All entries must be real values' # Having issues with this part...
    else:
        return array[mx.argmax(array)].item()

print(max_mx(a))
```

```python
def min_mx(array):
    """
    Returns the minimum value of an array.
    """
    if array.dtype == 'mlx.core.complex64':
        return 'Error: All entries must be real values' # Having issues with this part...
    else:
        return array[mx.argmin(array)].item()

print(min_mx(a))
```
